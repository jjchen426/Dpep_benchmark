```python
# 扫描/home/junjiechen/1_work/250401-Dpepalign/Benchmark/datasets/colabfold目录下的所有子目录中的msa_templ目录，将其中名称包含rank_001的pdb文件复制到/home/junjiechen/1_work/250401-Dpepalign/Benchmark/datasets/Colabfold/PDBs目录下
import os
import shutil

source_root = '/home/junjiechen/1_work/250401-Dpepalign/Benchmark/datasets/colabfold'
destination_root = '/home/junjiechen/1_work/250401-Dpepalign/Benchmark/datasets/Colabfold/PDBs'
if not os.path.exists(destination_root):
    os.makedirs(destination_root)

for dirpath, dirnames, filenames in os.walk(source_root):
    for dirname in dirnames:
        if dirname == 'msa_templ':
            msa_templ_path = os.path.join(dirpath, dirname)
            for dirpath, dirnames, filenames in os.walk(msa_templ_path):
                for filename in filenames:
                    if 'rank_001' in filename and filename.endswith('.pdb'):
                        source_file = os.path.join(dirpath, filename)
                        destination_file = os.path.join(destination_root, filename)
                        shutil.copy2(source_file, destination_file)
                        print(f'Copied: {source_file} to {destination_file}')
```
