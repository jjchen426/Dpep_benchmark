```python
import pdb

import numpy as np
from Bio import PDB
import copy
import os

def add_noise_to_pdb(input_pdb, output_pdb, noise_std):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure("struct", input_pdb)
    
    # 遍历所有原子
    for atom in structure.get_atoms():
        # 生成 3D 噪声向量
        noise = np.random.normal(0, noise_std, 3)
        # 叠加坐标
        atom.set_coord(atom.get_coord() + noise)
    io = PDB.PDBIO()
    io.set_structure(structure)
    io.save(output_pdb)

noise = 0.5
pdb_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/dataset/PepSet_AF3_noC_pass"
pdb_het_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/dataset/PepSet_AF3_noC_pass-HETATM"
noised_path = f"/home/junjiechen/1_work/250401-Dpepalign/Benchmark/dataset/PepSet_AF3_noC_pass-{noise}"
noised_het_path = f"/home/junjiechen/1_work/250401-Dpepalign/Benchmark/dataset/PepSet_AF3_noC_pass-HETATM-{noise}"
if not os.path.exists(noised_path):
    os.makedirs(noised_path, exist_ok=True)
if not os.path.exists(noised_het_path):
    os.makedirs(noised_het_path, exist_ok=True)

pdbs = os.listdir(pdb_path)

for file in pdbs:
    path = f'{pdb_path}/{file}'
    path_het = f'{pdb_het_path}/{file}'
    add_noise_to_pdb(path, f"{noised_path}/{file}", noise_std=noise)
    add_noise_to_pdb(path_het, f"{noised_het_path}/{file}", noise_std=noise)
```

```python
# 在指定目录中写入PDB.list文件，列出所有PDB文件的绝对路径
noise = 0.5
noised_path = f"/home/junjiechen/1_work/250401-Dpepalign/Benchmark/dataset/PepSet_AF3_noC_pass"
noised_het_path = f"/home/junjiechen/1_work/250401-Dpepalign/Benchmark/dataset/PepSet_AF3_noC_pass-HETATM"
with open(f"{noised_path}/PDB.list", "w") as f, open(f"{noised_het_path}/PDB.list", "w") as f_het:
    for file in pdbs:
        f.write(f"{os.path.abspath(f'{noised_path}/{file}')} \n")
        f_het.write(f"{os.path.abspath(f'{noised_het_path}/{file}')} \n")
```
