### 按照蛋白链对齐，计算多肽链骨架的RMSD

```python
import os
import json
import numpy as np
import pandas as pd
from Bio.PDB import PDBParser, FastMMCIFParser, Superimposer

with open('PDB.list', 'r') as f:
    pdbs = f.read().splitlines() 
# pdbs = [pdb for pdb in pdbs if pdb.startswith("2p1o") or pdb.startswith("2fka") or pdb.startswith("4zoz") or pdb.startswith("4x2h")]

pdbparser = PDBParser(QUIET=True)
mmcifparser = FastMMCIFParser(QUIET=True)

for pdb in pdbs:
    predicted_pdb_dir = f"./pro_nomsa_template-pep_nomsa_notemplate/output/{pdb}"
    ref_pdb_path = f"./dimer/{pdb}.pdb"
    if not os.path.exists(ref_pdb_path):
        raise FileNotFoundError(f"Reference PDB file not found: {ref_pdb_path}")
    if not os.path.isdir(predicted_pdb_dir):
        continue

    structure_ref = pdbparser.get_structure('ref', ref_pdb_path)
    chain_L_ref = structure_ref[0]['L']

    rows = []
    sample_dirs = [
        d for d in os.listdir(predicted_pdb_dir)
        if d.startswith('seed-') and '_sample-' in d and os.path.isdir(os.path.join(predicted_pdb_dir, d))
    ]

    # 指定42、43、44在sample dirs中
    sample_dirs = [d for d in sample_dirs if any(seed in d for seed in ['seed-42', 'seed-43', 'seed-44'])]

    for sample in sorted(sample_dirs):
        sample_dir = os.path.join(predicted_pdb_dir, sample)

        seed_str, sample_str = sample.split('_sample-')
        seed = int(seed_str.replace('seed-', ''))
        sample_id = int(sample_str)

        conf_json_path = os.path.join(sample_dir, f"{pdb}_{sample}_confidences.json")
        summary_json_path = os.path.join(sample_dir, f"{pdb}_{sample}_summary_confidences.json")
        cif_path = os.path.join(sample_dir, f"{pdb}_{sample}_model.cif")

        if not (os.path.exists(conf_json_path) and os.path.exists(summary_json_path) and os.path.exists(cif_path)):
            continue

        with open(conf_json_path, 'r') as f_json:
            conf_data = json.load(f_json)
        atom_chain_ids = conf_data.get('atom_chain_ids', [])
        atom_plddts = conf_data.get('atom_plddts', [])
        if len(atom_chain_ids) == 0 or len(atom_chain_ids) != len(atom_plddts):
            continue

        b_chain_plddt = [
            float(plddt)
            for chain_id, plddt in zip(atom_chain_ids, atom_plddts)
            if chain_id == 'B'
        ]
        if len(b_chain_plddt) == 0:
            continue
        pep_plddt = round(float(np.mean(b_chain_plddt)), 4)

        with open(summary_json_path, 'r') as f_json:
            summary_data = json.load(f_json)
        if 'iptm' not in summary_data or 'ranking_score' not in summary_data:
            continue
        iptm = round(float(summary_data['iptm']), 4)
        ranking_score = round(float(summary_data['ranking_score']), 4)

        structure_pred = mmcifparser.get_structure('pred', cif_path)
        try:
            chain_L_pred = structure_pred[0]['B']
        except KeyError:
            continue

        align_chain_ref = None
        for ch in structure_ref[0]:
            if ch.id != 'L':
                align_chain_ref = ch
                break
        if align_chain_ref is None:
            continue

        align_chain_pred = None
        for ch in structure_pred[0]:
            if ch.id != chain_L_pred.id:
                align_chain_pred = ch
                break
        if align_chain_pred is None:
            continue

        align_ref_atoms = [atom for atom in align_chain_ref.get_atoms() if atom.get_id() in ['N', 'CA', 'C', 'O']]
        align_pred_atoms = [atom for atom in align_chain_pred.get_atoms() if atom.get_id() in ['N', 'CA', 'C', 'O']]
        if len(align_ref_atoms) == 0 or len(align_pred_atoms) == 0:
            continue
        if len(align_ref_atoms) != len(align_pred_atoms):
            continue

        sup = Superimposer()
        try:
            sup.set_atoms(align_ref_atoms, align_pred_atoms)
        except Exception:
            continue
        rot, tran = sup.rotran

        ref_L_atoms = [atom for atom in chain_L_ref.get_atoms() if atom.get_id() in ['N', 'CA', 'C', 'O']]
        pred_L_atoms = [atom for atom in chain_L_pred.get_atoms() if atom.get_id() in ['N', 'CA', 'C', 'O']]
        if len(ref_L_atoms) == 0 or len(pred_L_atoms) == 0:
            continue
        if len(ref_L_atoms) != len(pred_L_atoms):
            continue

        ref_coords = np.array([a.get_coord() for a in ref_L_atoms], dtype=float)
        pred_coords = np.array([a.get_coord() for a in pred_L_atoms], dtype=float)
        pred_coords_aligned = pred_coords @ rot + tran

        diffs = ref_coords - pred_coords_aligned
        scRMSD = round(float(np.sqrt(np.mean(np.sum(diffs**2, axis=1)))), 4)

        rows.append(
            (
                pdb,
                seed,
                sample_id,
                pep_plddt,
                iptm,
                ranking_score,
                scRMSD,
            )
        )

    result = pd.DataFrame(
        rows,
        columns=['complex', 'seed', 'id', 'pep_plddt', 'iptm', 'ranking_score', 'scRMSD']
    )
    result.to_csv(f"{predicted_pdb_dir}/metrics_summary.csv", index=False)

    result_filtered = result[result['scRMSD'] <= 2.5].reset_index(drop=True)
    result_filtered.to_csv(f"{predicted_pdb_dir}/metrics_filtered_scRMSD_le25.csv", index=False)
    result_filtered
```

```python
import os
import pandas as pd

cutoffs = [1, 1.5, 2, 2.5]

pep_plddt_cutoff = 70
iptm_cutoff = 0.70

str_plddt_cutoff = str(pep_plddt_cutoff)
str_iptm_cutoff = "0" + str(int(100 * iptm_cutoff))

with open('/home/junjiechen/1_work/250401-Dpepalign/Benchmark/alphafold/PepSet_dimer_withC/PepSet_dimer_withC.list', 'r') as f:
    pdbs = [line.strip() for line in f if line.strip()]

pdb_w = max(4, max(len(p) for p in pdbs))
col_w = 8

def format_block(values):
    return " ".join([str(v).rjust(col_w) for v in values])

header1 = f"{'PDB':<{pdb_w}}  " + format_block([f"{c:g}" for c in cutoffs])
print(header1)

csv_rows = []
csv_header = ["pdb"] + [f"cutoff_{c:g}" for c in cutoffs]
csv_rows.append(csv_header)

for pdb in pdbs:
    df_path = f'./PepSet_dimer_withC/output/{pdb}/metrics_summary-seed42_43_44.csv'
    row = [pdb]

    if not os.path.exists(df_path):
        row.extend(["NA"] * len(cutoffs))
        csv_rows.append(row)
        print(f"{pdb:<{pdb_w}}  " + format_block(["NA"] * len(cutoffs)))
        continue

    df = pd.read_csv(df_path)
    if df.empty or 'complex' not in df.columns:
        row.extend(["NA"] * len(cutoffs))
        csv_rows.append(row)
        print(f"{pdb:<{pdb_w}}  " + format_block(["NA"] * len(cutoffs)))
        continue

    oricount = df['complex'].nunique()
    if oricount == 0:
        row.extend(["NA"] * len(cutoffs))
        csv_rows.append(row)
        print(f"{pdb:<{pdb_w}}  " + format_block(["NA"] * len(cutoffs)))
        continue

    rates = []
    for cutoff in cutoffs:
        count = df[
            (df['pep_plddt'] >= pep_plddt_cutoff)
            & (df['iptm'] >= iptm_cutoff)
            & (df['scRMSD'] <= cutoff)
        ]['complex'].nunique()
        rate = count / oricount * 100
        rate_str = f"{rate:.2f}"
        rates.append(rate_str)
        row.append(rate_str)

    csv_rows.append(row)
    print(f"{pdb:<{pdb_w}}  " + format_block(rates))

df_out = pd.DataFrame(csv_rows[1:], columns=csv_rows[0])
mean_series = df_out.drop(columns=['pdb']).replace('NA', pd.NA).astype(float).mean()
mean_row = ['MEAN'] + [f"{mean_series[col]:.2f}" if pd.notna(mean_series[col]) else "NA" for col in df_out.columns[1:]]
csv_rows.append(mean_row)

print(f"{'MEAN':<{pdb_w}}  " + format_block(mean_row[1:]))

csv_output_path = f'./PepSet_dimer_withC/output/success_rates_by_plddt{str_plddt_cutoff}_iptm{str_iptm_cutoff}-seed42_43_44.csv'
pd.DataFrame(csv_rows[1:], columns=csv_rows[0]).to_csv(csv_output_path, index=False)
print(f"CSV saved to: {csv_output_path}")
```

```python
# 将成功预测的pdb中scRMSD最小的pdb复制到一个新的文件夹中
with open('/home/junjiechen/1_work/250401-Dpepalign/Benchmark/alphafold/PepSet_dimer_withC/PepSet_dimer_withC.list', 'r') as f:
    pdbs = [line.strip() for line in f if line.strip()]

with open('./PepSet_dimer_withC/results-3seeds.txt', 'w') as result_file:
    result_file.write("pdb,seed,sample_id,scRMSD,iptm,pep_plddt\n")
    for pdb in pdbs:
        predicted_pdb_dir = f"./PepSet_dimer_withC/output/{pdb}"
        if not os.path.isdir(predicted_pdb_dir):
            continue

        df_path = f"{predicted_pdb_dir}/metrics_summary-seed42_43_44.csv"
        if not os.path.exists(df_path):
            continue

        df = pd.read_csv(df_path)
        if df.empty or 'scRMSD' not in df.columns:
            continue

        df_filtered = df[
            (df['scRMSD'] <= 2.5) & (df['iptm'] >= 0.70) & (df['pep_plddt'] >= 70)
        ]
        if df_filtered.empty:
            continue

        best_row = df_filtered.loc[df_filtered['scRMSD'].idxmin()]
        best_seed = int(best_row['seed'])
        best_sample_id = int(best_row['id'])

        source_cif = f"{predicted_pdb_dir}/seed-{best_seed}_sample-{best_sample_id}/{pdb}_seed-{best_seed}_sample-{best_sample_id}_model.cif"
        target_dir = f"./PepSet_dimer_withC/best_models-3seeds"
        os.makedirs(target_dir, exist_ok=True)
        target_cif = f"{target_dir}/{pdb}.cif"

        if os.path.exists(source_cif):
            os.system(f"cp {source_cif} {target_cif}")
        result_file.write(f"{pdb},{best_seed},{best_sample_id},{float(best_row['scRMSD'])},{float(best_row['iptm'])},{float(best_row['pep_plddt'])}\n")
```
