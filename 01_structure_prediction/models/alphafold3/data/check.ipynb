```python
# 检查summary文件，计算每个样本的成功率。成功率的定义为每个complex存在一个多肽plddt>=70，iptm>=0.7，多肽rmsd<2.5的样本，则这个样本预测成功
import pandas as pd
df = pd.read_csv('/home/junjiechen/1_work/250401-Dpepalign/Benchmark/01_structure_prediction/models/alphafold3/data/summary/msa_template/metrics_summary.csv')
success_count = 0
successful_complexes = []
for complex_name, group in df.groupby('complex'):
    if any(
        (group['pep_plDDT'] >= 70) &
        (group['ipTM'] >= 0.7) &
        (group['scRMSD'] < 2.5)
    ):
        success_count += 1
        successful_complexes.append(complex_name)

total_count = df['complex'].nunique()
success_rate = success_count / total_count if total_count > 0 else 0

print(f"Oracle Success count: {success_count}")
print(f"Total count: {total_count}")
print(f"Oracle Success rate: {success_rate:.2%}")

# 将满足条件的样本的complex名称保存到一个新的txt文件中
with open('/home/junjiechen/1_work/250401-Dpepalign/Benchmark/01_structure_prediction/models/alphafold3/data/summary/msa_template/successful_complexes.txt', 'w') as f:
    for complex_name in sorted(set(successful_complexes)):
        f.write(f"{complex_name}\n")
```
