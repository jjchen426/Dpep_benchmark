```python
import json
import os

# 读取job-add-msa.json文件，将其中所有蛋白链的MSA路径提取出来，将路径对应的目录复制到MSA_pro_all目录下
with open('job-add-msa.json', 'r') as f:
    jobs = json.load(f)
for job in jobs:
    name = job['name']
    msa_dir = job['sequences'][0]['proteinChain']['msa']['precomputed_msa_dir']
    os.system(f'cp -r {msa_dir} ./home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_filter/datasets/PepSet-passed-dimer_processed/MSA_pro_all/{name}')
    print(f'Job name: {name}, MSA dir: {msa_dir}')
            
```

```python
import json

with open('job-add-msa.json', 'r') as f:
    jobs = json.load(f)

count_exist = 0
count_error = 0
for job in jobs:
    msa_dir = job['sequences'][0]['proteinChain']['msa']['precomputed_msa_dir']
    if "MSA_pro" in msa_dir:
        count_exist += 1
    else:
        count_error += 1
print(f"Count of existing MSA_pro directories: {count_exist}")
print(f"Count of errors: {count_error}")
```

```python
import os
with open('./dimer_filtered/PDB.list', 'r') as f:
    pdbs = [line.strip() for line in f.readlines()]
count_found = 0
count_not_found = 0
for pdb in pdbs:
    pdb_id = pdb.split("_")[0]
    if pdb_id in os.listdir('MSA_pro'):
        count_found += 1
    else:
        count_not_found += 1
print(f"Count of found PDB IDs in MSA_pro: {count_found}")
print(f"Count of not found PDB IDs in MSA_pro: {count_not_found}")
```
