```python
import os
import re
import subprocess
from pathlib import Path
import pandas as pd
from Bio.PDB import PDBParser, MMCIFParser
```

```python


with open('/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_filter/datasets/PepSet-passed-dimer_processed/PDB.list', 'r') as f:
    pdbs = [pdb.strip() for pdb in f.readlines()]

origin_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/protenix/PepSet/dimer_filtered"

with open('jobs.list', 'w') as f:
    for pdb in pdbs:
        pred_path = f'/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_filter/Protenix/2000-0.1T-dimer/{pdb}/ligandmpnn_seed42'
        count = 0
        for dir in os.listdir(pred_path):
            if dir.split('_')[-1].isdigit() and dir.startswith(pdb + "_"):
                count += 1
        for i in range(1, count+1):
            model = pdb + "_" + str(i)
            for seed in ['seed_42', 'seed_43', 'seed_44']:
                for j in range(5):
                    sample_path = f'{pred_path}/{model}/{seed}/predictions/{model}_sample_{j}.cif'
                    dockq_cmd = f'DockQ {sample_path} {origin_path}/{pdb}.pdb --short --allowed_mismatches 20 --mapping AB:AB\n'
                    f.write(dockq_cmd)
```
