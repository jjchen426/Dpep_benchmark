```python
import json
import os

with open('/home/junjiechen/1_work/250401-Dpepalign/Benchmark-Filter/datasets/PepSet_3per_length/PDB.list', 'r') as f:
    pdbs = [line.strip() for line in f.readlines()]
with open(f'msa_path.txt', 'w') as f_out:
    for pdb in pdbs:
        print(pdb)
        json_path = f'/home/junjiechen/1_work/250401-Dpepalign/Benchmark-Filter/Protenix/10000-0.1T/{pdb}/protenix_pred-filter1-add-msa.json'
        with open(json_path, 'r') as f:
            data = json.load(f)
        # 将第一个job的protein的msa路径保存到txt文件中
            msa_path = data[0]['sequences'][0]['proteinChain']['msa']['precomputed_msa_dir']
            pairing_db = data[0]['sequences'][0]['proteinChain']['msa']['pairing_db']
            f_out.write(f'{msa_path},{pairing_db}\n')
        
```

```python
import os

with open('msa_path.txt', 'r') as f:
    msa_lines = [line.strip().split(',') for line in f.readlines()]

for path, _ in msa_lines:
    pro = path.split('/')[-3].split('_')[0]
    # 将line的路径软链接到当前路径下，修改名称为pro
    os.system(f'cp -r {path} ./{pro}')
```
