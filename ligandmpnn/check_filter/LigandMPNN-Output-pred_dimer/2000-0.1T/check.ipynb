```python
import os
noC = 93
withC = 0
for roots, dirs, files in os.walk('./2000-0.1T'):
    for file in files:
        if file.endswith('.fa'):
            check = False
            filepath = os.path.join(roots, file)
            with open(filepath, 'r') as f:
                lines = f.readlines()
            for i, line in enumerate(lines):
                if line.startswith('>'):
                    continue
                else:
                    if 'C' in line:
                        print(f'File: {filepath}, Line {i+1} contains Cysteine (C).')
                        withC += 1
                        check = True
                        break
            if not check:
                noC -= 1
print(f'Total files with Cysteine (C): {withC}')
print(f'Total files without Cysteine (C): {noC}')
```
