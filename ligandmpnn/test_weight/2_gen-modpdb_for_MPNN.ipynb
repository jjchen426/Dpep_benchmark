```python
# 生成pdb路径文件
import os

dir_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_weight/datasets/PepSet-noise-0.5"
with open(f'{dir_path}/PDB.list', 'w') as f:
    for root, dirs, files in os.walk(f'{dir_path}/Merged_PDBs'):
        for file in files:
            if file.endswith('.pdb'):
                full_path = os.path.join(root, file)
                print(full_path)
                f.write(full_path + '\n')

# 编写json文件，格式为{"路径"：""},其中路径为complex.list中每一行的内容，去掉换行符
import json
json_dict = {}
with open(f"{dir_path}/PDB.list", 'r') as f:
    for line in f:
        path = line.strip()
        json_dict[path] = ""
with open(f"{dir_path}/Merged.json", "w") as json_file:
    json.dump(json_dict, json_file, indent=4)

# 将path中的Merged_PDBs替换为Processed_PDBs，写入新的json文件
processed_json_dict = {}
with open(f"{dir_path}/Merged.json", 'r') as f:
    data = json.load(f)
    for key in data.keys():
        new_key = key.replace("Merged_PDBs", "Processed_PDBs")
        processed_json_dict[new_key] = ""
with open(f"{dir_path}/Processed.json", "w") as json_file:
    json.dump(processed_json_dict, json_file, indent=4)
```

```python
import json
# 读取Merged.json，将其中包含2p1o、4x2h、4zoz的路径写入新的json文件
base_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_weight/datasets/PepSet-noise-0.5"
filtered_json_dict = {}
filtered_json_dict2 = {}
with open(f"{base_path}/Merged.json", 'r') as f, open(f"{base_path}/Processed.json", 'r') as f2:
    data = json.load(f)
    data2 = json.load(f2)
    for key in data.keys():
        if "2p1o" in key or "4x2h" in key or "4zoz" in key or "2fka" in key:
            filtered_json_dict[key] = ""
    for key in data2.keys():
        if "2p1o" in key or "4x2h" in key or "4zoz" in key or "2fka" in key:
            filtered_json_dict2[key] = ""
with open(f"{base_path}/fix_pdb.json", "w") as json_file:
    json.dump(filtered_json_dict, json_file, indent=4)
with open(f"{base_path}/Processed-fix_pdb.json", 'w') as json2:
    json.dump(filtered_json_dict2, json2, indent=4)
```
