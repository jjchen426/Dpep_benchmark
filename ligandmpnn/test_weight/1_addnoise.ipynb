```python
import numpy as np
from Bio import PDB
import copy
import os

def add_noise_to_pdb(input_pdb, output_pdb, noise_std):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure("struct", input_pdb)
    
    # 遍历所有原子
    for atom in structure.get_atoms():
        # 生成 3D 噪声向量
        noise = np.random.normal(0, noise_std, 3)
        # 叠加坐标
        atom.set_coord(atom.get_coord() + noise)
    io = PDB.PDBIO()
    io.set_structure(structure)
    io.save(output_pdb)

noise = 0.4
pdb_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_weight/datasets/fix_pdb.list"
noised_path = f"/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_weight/datasets/PepSet-noise-{noise}/Merged_PDBs"
if not os.path.exists(noised_path):
    os.makedirs(noised_path, exist_ok=True)

with open(pdb_path, "r") as f:
    pdbs = [line.strip() for line in f.readlines()]

for file in pdbs:
    path = f'/home/junjiechen/1_work/250401-Dpepalign/Benchmark/ligandmpnn/test_weight/datasets/PepSet/Merged_PDBs/{file}.pdb'
    add_noise_to_pdb(path, f"{noised_path}/{file}.pdb", noise_std=noise)
```
