```python
# 读取PepSet以及添加噪声的六个目录下的seq_rec_values.csv文件，将六个文件按照列合并到新的DataFrame中，并保存为combined_seq_rec_values.csv文件
import pandas as pd
data_sets = ["PepSet", "PepSet-noise-0.1", "PepSet-noise-0.2", "PepSet-noise-0.3", "PepSet-noise-0.4", "PepSet-noise-0.5"]
df_list = []
for i, data_set in enumerate(data_sets):
    df = pd.read_csv(f"./outputs/{data_set}/seq_rec_values-{i}-dimer.csv")
    df_list.append(df)

# 按列交错合并：第1~6列取各df的第1列，第7~12列取各df的第2列，以此类推，同时重命名列名为Method-res-noiselevel
combined_df = pd.DataFrame()
num_methods = df_list[0].shape[1]
for j in range(num_methods):
    for i, data_set in enumerate(data_sets):
        col_name = f"{df_list[i].columns[j]}-{data_set.split('-')[-1]}"
        combined_df[col_name] = df_list[i].iloc[:, j]
        
# 转置combined_df, 并保存为新的CSV文件
transposed_df = combined_df.T
# 计算每一行的均值，保存到新的一列"Average"
transposed_df['Average'] = transposed_df.mean(axis=1)
transposed_df.to_csv("./outputs/pepset-dimer-combined_seq_rec_values_transposed.csv", index=True)

```
