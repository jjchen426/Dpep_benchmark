```python
# 检查指定路径下的所有pdb的B链是否存在CYS
import os
from Bio.PDB import PDBParser, PDBIO, FastMMCIFParser

pdb_path = "/home/junjiechen/1_work/250401-Dpepalign/Benchmark/alphafold3/PepSet_dimer_withC/best_models-3seeds"
for file in os.listdir(pdb_path):
    if file.endswith(".cif"):
        parser = FastMMCIFParser(QUIET=True)
        structure = parser.get_structure(file[:-4], os.path.join(pdb_path, file))
        for model in structure:
            for chain in model:
                if chain.id == 'B':
                    for residue in chain:
                        if residue.get_resname() == 'CYS':
                            print(f"{file} has CYS in chain B")
```
